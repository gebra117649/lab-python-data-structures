## Solution: Managing Customer Orders

As part of a business venture, you are starting an online store that sells various products. To ensure smooth operations, you need to develop a program that manages customer orders and inventory.

````python
## Exercise: Managing Customer Orders

# 1. Define a list called `products` that contains the following items: "t-shirt", "mug", "hat", "book", "keychain".

products = ["t-shirt", "mug", "hat", "book", "keychain"]
print(f"Products: {products}\n")

# 2. Create an empty dictionary called `inventory`.

inventory = {}
print(f"Inventory: {inventory}\n")

# 3. Ask the user to input the quantity of each product available in the inventory. 
# Use the product names from the `products` list as keys in the `inventory` dictionary and assign the respective quantities as values.

inventory["t-shirt"] = int(input("Enter quantity for t-shirt: "))
inventory["mug"] = int(input("Enter quantity for mug: "))
inventory["hat"] = int(input("Enter quantity for hat: "))
inventory["book"] = int(input("Enter quantity for book: "))
inventory["keychain"] = int(input("Enter quantity for keychain: "))

print(f"Inventory: {inventory}\n")

# 4. Create an empty set called `customer_orders`.

customer_orders = set()

# 5. Ask the user to input the name of three products that a customer wants to order 
# (from those in the products list, meaning three products out of "t-shirt", "mug", "hat", "book" or "keychain". 
# Add each product name to the `customer_orders` set.

order1 = input("Enter the first product to order: ")
customer_orders.add(order1)

order2 = input("Enter the second product to order: ")
customer_orders.add(order2)

order3 = input("Enter the third product to order: ")
customer_orders.add(order3)

# !!!!NOTE!!!
# # Edge case not handled yet: customer input is not in the product list!
# proposed solution: check with "if"
# if order1 in products:
#   customer_orders.add(order1)
# else:
#   print(f"Sorry, {order1} is not available in our store!")

# 6. Print the products in the `customer_orders` set.

print(f"Customer orders: {customer_orders}\n")

# 7. Calculate the following order statistics:
# - Total Products Ordered: The total number of products in the `customer_orders` set.
total_products_ordered = len(customer_orders)
print(f"Total products ordered: {total_products_ordered}\n")

# - Percentage of Products Ordered: The percentage of products ordered compared to the total available products.
percentage_ordered = (total_products_ordered / len(products)) * 100
print(f"Percentage ordered:{percentage_ordered}\n")

# Store these statistics in a tuple called `order_status`.

order_status = (total_products_ordered, percentage_ordered)

# 8. Print the order statistics using the following format:
   # ```
   # Order Statistics:
   # Total Products Ordered: <total_products_ordered>
   # Percentage of Products Ordered: <percentage_ordered>% 
   # ```

print("Order Statistics:")
print(f"Total Products Ordered: {total_products_ordered}")
print(f"Percentage of Products Ordered: {percentage_ordered}%\n" )    

# 9. Update the inventory by subtracting 1 from the quantity of each product. Modify the `inventory` dictionary accordingly.

inventory["t-shirt"] -= 1
inventory["mug"] -= 1
inventory["hat"] -= 1
inventory["book"] -= 1
inventory["keychain"] -= 1

# 10. Print the updated inventory, displaying the quantity of each product on separate lines.
print(f"The quantity of t-shirt is: {inventory['t-shirt']}")
print(f"The quantity of mug is: {inventory['mug']}")
print(f"The quantity of hat is: {inventory['hat']}")
print(f"The quantity of book is: {inventory['book']}")
print(f"The quantity of keychain is: {inventory['keychain']}")

# !!!!NOTE!!!
# Edge cases not handled 
# for example the quantity is a minus

# Solve the exercise by implementing the steps using the Python concepts of lists, dictionaries, sets, and basic input/output operations. 
````
